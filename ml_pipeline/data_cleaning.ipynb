```python
import numpy as np 
import pandas as pd
```

```python
# Load the dataset

df = pd.read_excel('../data/extention of Z-Alizadeh sani dataset.xlsx')

df.head()
```

```python
df.info()
print("shape",df.shape)
```

```python
# Check for any missing (empty) values in the entire dataset

total_missing = df.isnull().sum().sum()
print("Total Missing Values:", total_missing)

# check for any exact duplicate rows 

total_duplicates = df.duplicated().sum()
print("Total Duplicate Rows:", total_duplicates)
```

```python
print(df.describe().T)
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns

# Loop through each categorical column and print its name along with its unique values
for col in categorical_cols:
    unique_values = df[col].unique()
    print(f"{col}: {unique_values}")
```
