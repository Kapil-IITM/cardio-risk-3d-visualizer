```python
YN_COLS = ["Obesity","CRF","CVA","Airway disease","Thyroid Disease","CHF","DLP",
           "Weak Peripheral Pulse","Lung rales","Systolic Murmur","Diastolic Murmur",
           "Dyspnea","Atypical","Nonanginal","LowTH Ang","LVH","Poor R Progression"]
DROP_COLS = ["Cath","LAD","LCX","RCA","Exertional CP","Weight","Length"]
```

```python
def preprocess(data):
    X = data.drop(columns=[c for c in DROP_COLS if c in data.columns]).copy()

    X["Sex"] = X["Sex"].map({"Male": 1, "Female": 0})
    for c in YN_COLS:
        X[c] = X[c].map({"Y": 1, "N": 0})
    X["VHD"] = X["VHD"].str.lower().map({"n": 0, "mild": 1, "moderate": 2, "severe": 3})

    # Categories fix rakho, taaki backend mein bhi same columns bane
    X["BBB"] = pd.Categorical(X["BBB"], categories=["N", "LBBB", "RBBB"])
    X["Region RWMA"] = pd.Categorical(X["Region RWMA"], categories=[0, 1, 2, 3, 4])
    X = pd.get_dummies(X, columns=["BBB", "Region RWMA"], dtype=int)

    assert X.isnull().sum().sum() == 0, "Mapping mein spelling mismatch"
    return X
```

```python
X = preprocess(df)
y = pd.DataFrame({t: (df[t] == pos_label[t]).astype(int) for t in TARGETS})

print("X shape:", X.shape)
print(y.mean().round(3))
print("Leakage columns X mein:", [c for c in TARGETS if c in X.columns])
print("Exertional CP/Weight/Length X mein:", [c for c in ["Exertional CP","Weight","Length"] if c in X.columns])
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline

rng = np.random.RandomState(0)
noise = pd.DataFrame(rng.randn(303, 500))          # 500 random columns, koi signal nahi
yt = y["Cath"]
skf = StratifiedKFold(5, shuffle=True, random_state=0)

# GALAT: poore data par top-20 chuno, phir CV
top = SelectKBest(f_classif, k=20).fit(noise, yt).get_support()
wrong = cross_val_score(LogisticRegression(max_iter=1000), noise.loc[:, top], yt, cv=skf, scoring="roc_auc")

# SAHI: selection Pipeline ke andar
right = cross_val_score(
    make_pipeline(SelectKBest(f_classif, k=20), LogisticRegression(max_iter=1000)),
    noise, yt, cv=skf, scoring="roc_auc")

print("Galat tarika AUC:", wrong.mean().round(3))
print("Sahi tarika AUC :", right.mean().round(3))
```

```python
import warnings; warnings.filterwarnings("ignore")
from sklearn.model_selection import RepeatedStratifiedKFold, cross_validate
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.dummy import DummyClassifier
```

```python
SEED = 42
CV = RepeatedStratifiedKFold(n_splits=5, n_repeats=3, random_state=SEED)
SCORING = ["roc_auc", "accuracy", "precision", "recall", "f1"]
results = []

def evaluate(name, model, X, target):
    s = cross_validate(model, X, y[target], cv=CV, scoring=SCORING, n_jobs=-1)
    row = {"model": name, "target": target}
    for m in SCORING:
        row[m] = s[f"test_{m}"].mean()
        row[m + "_std"] = s[f"test_{m}"].std()
    results.append(row)
    print(f"{name:14s} {target:5s} AUC={row['roc_auc']:.3f} (+/-{row['roc_auc_std']:.3f})  "
          f"Recall={row['recall']:.3f}  F1={row['f1']:.3f}  Acc={row['accuracy']:.3f}")
```

```python
for t in TARGETS:
    evaluate("Dummy", DummyClassifier(strategy="prior"), X, t)
```

```python
lr = make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000))
rf_plain = RandomForestClassifier(n_estimators=300, min_samples_leaf=3, random_state=0)

for t in TARGETS:
    evaluate("LogReg", lr, X, t)
    evaluate("RF", rf_plain, X, t)
    print("-" * 75)
```

```python
res = pd.DataFrame(results)
print("=== ROC-AUC ===")
display(res.pivot_table(index="model", columns="target", values="roc_auc")[TARGETS].round(3))
print("=== Recall ===")
display(res.pivot_table(index="model", columns="target", values="recall")[TARGETS].round(3))
```

```python
lr_bal = make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000, class_weight="balanced"))
rf_bal = RandomForestClassifier(n_estimators=300, min_samples_leaf=3,
                                class_weight="balanced", random_state=0)

for t in TARGETS:
    evaluate("LogReg_bal", lr_bal, X, t)
    evaluate("RF_bal", rf_bal, X, t)
    print("-" * 75)
```

```python
res = pd.DataFrame(results)
for metric in ["roc_auc", "recall", "f1"]:
    print(f"=== {metric} ===")
    display(res.pivot_table(index="model", columns="target", values=metric)[TARGETS].round(3))
```

```python
drop_sets = {
    "base":         [],
    "no_Obesity":   ["Obesity"],
    "no_Neut":      ["Neut"],
    "no_Lymph":     ["Lymph"],
    "no_ultrarare": ["CHF", "LowTH Ang"],
    "all_drops":    ["Obesity", "Neut", "CHF", "LowTH Ang"],
}

for t in TARGETS:
    for name, cols in drop_sets.items():
        Xd = X.drop(columns=cols)
        evaluate(f"RF_{name}", rf_bal, Xd, t)
        evaluate(f"LR_{name}", lr_bal, Xd, t)
    print("-" * 75)
```

```python
EXTRA_DROPS = ["Obesity", "Neut", "CHF", "LowTH Ang"]
X_final = X.drop(columns=EXTRA_DROPS)
print(X_final.shape)         # (303, 54) aana chahiye
print("Dropped:", EXTRA_DROPS)
```

```python
from sklearn.calibration import CalibratedClassifierCV
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import brier_score_loss, roc_auc_score

skf = StratifiedKFold(5, shuffle=True, random_state=SEED)

def rf_final():
    return RandomForestClassifier(n_estimators=300, min_samples_leaf=3,
                                  class_weight="balanced", random_state=0)

rows = []
for t in TARGETS:
    base_rate = y[t].mean()
    rows.append({"target": t, "method": "baseline", "Brier": base_rate * (1 - base_rate), "AUC": 0.5})
    for name, m in {"raw": rf_final(),
                    "sigmoid": CalibratedClassifierCV(rf_final(), method="sigmoid", cv=5),
                    "isotonic": CalibratedClassifierCV(rf_final(), method="isotonic", cv=5)}.items():
        p = cross_val_predict(m, X_final, y[t], cv=skf, method="predict_proba", n_jobs=-1)[:, 1]
        rows.append({"target": t, "method": name,
                     "Brier": brier_score_loss(y[t], p), "AUC": roc_auc_score(y[t], p)})

cal = pd.DataFrame(rows)
print("=== Brier (low = good) ===")
display(cal.pivot_table(index="method", columns="target", values="Brier")[TARGETS].round(3))
print("=== AUC ===")
display(cal.pivot_table(index="method", columns="target", values="AUC")[TARGETS].round(3))
```

```python
import shap

explain_rf, explainers = {}, {}
for t in TARGETS:
    explain_rf[t] = rf_final().fit(X_final, y[t])
    explainers[t] = shap.TreeExplainer(explain_rf[t])
    joblib.dump(explain_rf[t], f"models/explain_rf_{t}.pkl")

def shap_pos(t, Xrow):
    """Disease wali class ke liye SHAP values, shape (patients, features)."""
    sv = np.array(explainers[t].shap_values(Xrow))
    if sv.ndim == 3:
        sv = sv[..., 1] if sv.shape[-1] == 2 else sv[1]
    return sv

print(shap_pos("Cath", X_final.iloc[:2]).shape)   # (2, 54) aana chahiye
```

```python
fig, axes = plt.subplots(1, 4, figsize=(24, 7))
for ax, t in zip(axes, TARGETS):
    sv = shap_pos(t, X_final)
    imp = pd.Series(np.abs(sv).mean(axis=0), index=X_final.columns).sort_values().tail(12)
    imp.plot(kind="barh", ax=ax)
    ax.set_title(f"{t}: top-12 mean |SHAP|")
plt.tight_layout(); plt.savefig("figures/shap_global.png", dpi=110); plt.show()
```

```python
import json

INPUT_FIELDS = [c for c in df.columns if c not in DROP_COLS + EXTRA_DROPS]
print(len(INPUT_FIELDS))          # 48 aana chahiye

schema = []
for c in INPUT_FIELDS:
    s = df[c]
    if s.dtype == object or str(s.dtype) == "str":
        item = {"name": c, "type": "choice", "values": sorted(s.unique().tolist()),
                "default": s.mode()[0]}
    elif s.nunique() == 2 and set(s.unique()) <= {0, 1}:
        item = {"name": c, "type": "binary", "values": [0, 1], "default": int(s.mode()[0])}
    elif c in ["Region RWMA", "Function Class"]:
        item = {"name": c, "type": "choice", "values": sorted(int(v) for v in s.unique()),
                "default": int(s.mode()[0])}
    else:
        item = {"name": c, "type": "number", "min": float(s.min()), "max": float(s.max()),
                "default": float(s.median())}
    schema.append(item)

json.dump(schema, open("models/form_schema.json", "w"), indent=1)

sample = json.loads(df.iloc[[0]][INPUT_FIELDS].to_json(orient="records"))[0]
print(json.dumps({"patient": sample}, indent=1))
print(sorted(os.listdir("models")))
```
